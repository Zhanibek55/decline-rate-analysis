```python
import sys
sys.path.append('.')  # Добавляем текущую директорию в путь

# Импортируем функции из вашего проекта
from utils import load_data, calculate_monthly_decline, plot_decline_trend, safe_to_pandas
from datetime import datetime
import polars as pl
import pandas as pd
import traceback

# Функция для подробной проверки данных
def debug_data(data, name):
    print(f"\n===== Отладка {name} =====")
    print(f"Тип: {type(data)}")
    
    if isinstance(data, (pl.DataFrame, pd.DataFrame)):
        print(f"Размер: {data.shape}")
        print(f"Колонки: {data.columns}")
        if isinstance(data, pl.DataFrame):
            print(f"Типы данных: {data.dtypes}")
        else:
            print(f"Типы данных: {data.dtypes}")
        
        # Проверка на наличие пустых значений
        if isinstance(data, pl.DataFrame):
            null_counts = {col: data.select(pl.col(col).is_null().sum()).item() for col in data.columns}
        else:
            null_counts = data.isnull().sum().to_dict()
        print(f"Пустые значения: {null_counts}")
        
        # Проверка колонок с датами
        date_cols = [col for col in data.columns if 'date' in col.lower()]
        if date_cols:
            print(f"Колонки с датами: {date_cols}")
            for col in date_cols:
                if isinstance(data, pl.DataFrame):
                    sample = data.select(pl.col(col)).head(3)
                else:
                    sample = data[col].head(3)
                print(f"  Примеры значений в {col}: {sample}")
    
    elif isinstance(data, (list, tuple)):
        print(f"Длина: {len(data)}")
        if len(data) > 0:
            print(f"Тип первого элемента: {type(data[0])}")
            if len(data) > 3:
                print(f"Первые 3 элемента: {data[:3]}")
            else:
                print(f"Все элементы: {data}")
    
    elif isinstance(data, (str, int, float, bool, datetime)):
        print(f"Значение: {data}")
    
    print("=" * 40)

# Загружаем тестовые данные
print("Загрузка данных...")
try:
    prod_df, events_df = load_data()
    debug_data(prod_df, "prod_df")
    debug_data(events_df, "events_df")
except Exception as e:
    print(f"Ошибка при загрузке данных: {e}")
    traceback.print_exc()

# Параметры расчета
start_date = datetime.strptime("2024-01-01", "%Y-%m-%d").date()
end_date = datetime.strptime("2024-12-01", "%Y-%m-%d").date()
window_size = 3  # Размер окна в месяцах
group_by_cols = ["horizon"]
rate_col = "oil_rate"

debug_data(start_date, "start_date")
debug_data(end_date, "end_date")
debug_data(window_size, "window_size")
debug_data(group_by_cols, "group_by_cols")
debug_data(rate_col, "rate_col")

# Проверяем, что колонки существуют в данных
required_cols = ["date", "uwi", rate_col] + group_by_cols
missing_cols = [col for col in required_cols if col not in prod_df.columns]
if missing_cols:
    print(f"ОШИБКА: В данных добычи отсутствуют колонки: {missing_cols}")
else:
    print("Все необходимые колонки присутствуют в данных добычи")

# Проверяем типы данных колонок
print("\nПроверка типов данных:")
for col in prod_df.columns:
    col_type = prod_df.select(pl.col(col)).dtypes[0]
    print(f"  {col}: {col_type}")
    
    # Особая проверка для колонки с датами
    if col == "date":
        sample_dates = prod_df.select(pl.col("date")).head(5)
        print(f"  Примеры дат: {sample_dates}")

# Проверяем генерацию пар дат для скользящего окна
print("\nГенерация пар дат для скользящего окна:")
dates = []
current_date = start_date.replace(day=1)
while current_date <= end_date:
    dates.append(current_date)
    # Переходим к следующему месяцу
    if current_date.month == 12:
        current_date = current_date.replace(year=current_date.year + 1, month=1)
    else:
        current_date = current_date.replace(month=current_date.month + 1)

print(f"Сгенерировано {len(dates)} дат:")
for i, d in enumerate(dates[:5]):
    print(f"  {i+1}. {d} (тип: {type(d)})")
if len(dates) > 5:
    print(f"  ... и еще {len(dates)-5} дат")

date_pairs = []
for i in range(len(dates) - window_size):
    date_pairs.append((dates[i], dates[i + window_size]))

print(f"\nСформировано {len(date_pairs)} пар дат:")
for i, (t0, t1) in enumerate(date_pairs[:3]):
    print(f"  {i+1}. {t0} - {t1}")
if len(date_pairs) > 3:
    print(f"  ... и еще {len(date_pairs)-3} пар")

# Проверяем первую пару дат
if date_pairs:
    t0, t1 = date_pairs[0]
    print(f"\nПроверка расчета для первой пары дат: {t0} - {t1}")
    try:
        # Проверяем форматирование дат
        t0_str = t0.strftime("%Y-%m-%d")
        t1_str = t1.strftime("%Y-%m-%d")
        print(f"  Строковое представление: {t0_str} - {t1_str}")
        
        # Фильтруем данные по датам для проверки
        filtered_df = prod_df.filter(
            (pl.col("date") >= t0_str) & 
            (pl.col("date") <= t1_str)
        )
        print(f"  После фильтрации по датам: {filtered_df.shape[0]} записей")
        
        # Проверяем группировку
        group_col = group_by_cols[0]
        groups = filtered_df.select(pl.col(group_col)).unique()
        print(f"  Уникальные значения в колонке {group_col}: {groups.shape[0]} групп")
        
        # Проверяем наличие данных для расчета
        if filtered_df.shape[0] == 0:
            print("  ОШИБКА: Нет данных после фильтрации по датам")
        elif groups.shape[0] == 0:
            print(f"  ОШИБКА: Нет уникальных значений в колонке {group_col}")
        else:
            print("  Данные готовы для расчета")
    except Exception as e:
        print(f"  ОШИБКА при проверке данных: {e}")
        traceback.print_exc()

# Теперь попробуем выполнить расчет с подробным логированием
print("\nРасчет динамики темпа падения...")
try:
    monthly_decline = calculate_monthly_decline(
        prod_df=prod_df,
        events_df=events_df,
        start_date=start_date,
        end_date=end_date,
        window_size=window_size,
        group_by_cols=group_by_cols,
        rate_col=rate_col,
        include_events=False
    )
    
    debug_data(monthly_decline, "monthly_decline (результат)")
    
    if monthly_decline.shape[0] > 0:
        print("\nРасчет успешно выполнен!")
        
        # Преобразуем в pandas DataFrame для отображения
        pandas_df = safe_to_pandas(monthly_decline)
        
        # Отображаем результаты
        print("\nПервые 5 строк результатов:")
        print(pandas_df.head(5))
        
        # Создаем график
        print("\nПостроение графика динамики...")
        fig = plot_decline_trend(
            monthly_decline,
            group_col=group_by_cols[0],
            rate_col="decline_percent",
            title=f"Динамика темпа падения {rate_col} по {group_by_cols[0]}"
        )
        fig.show()
    else:
        print("ОШИБКА: Результат расчета - пустой DataFrame")
except Exception as e:
    print(f"ОШИБКА при расчете: {e}")
    traceback.print_exc()
    
```
